```python
import pandas as pd
import numpy as np

import seaborn as sns

df = sns.load_dataset("titanic")

# Performing first look ritual

df.head()
df.info()
df.describe()
df.dtypes

df
```

```python
# Counting missing columns

count = df.isnull().sum()
print(count)
```

```python
# Handeling Missing Values

## dropping: class, alive, who, embarked (repeating columns)
df = df.drop(columns=['class', 'alive', 'who', 'embarked'])

## filling: age, deck, embark_town
df['age'] = df['age'].fillna(df['age'].median())
df['deck'] = df['deck'].cat.add_categories('Unknown').fillna('Unknown')
df['embark_town'] = df['embark_town'].fillna('Unknown')

handled_count = df.isnull().sum()
print(handled_count)
```

```python
# Checking datatypes after handling missing values

df.dtypes
```

```python
# Encoding Text Values

df = pd.get_dummies(df, columns=['sex', 'embark_town', 'deck'], drop_first=True)

df
```

```python
# Scaling

from sklearn.preprocessing import StandardScaler

features = df.drop(columns=['survived'])
target = df['survived']
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)

print(scaled_features)
```

```python
# Spliting into train/test

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(scaled_features, target, test_size=0.2, random_state=42)

print('Trained Rows: ', len(x_train))
print('Tested Rows: ', len(x_test))
```
