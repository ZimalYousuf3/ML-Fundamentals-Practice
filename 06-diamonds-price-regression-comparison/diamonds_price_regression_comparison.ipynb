```python
import numpy as np
import pandas as pd
import seaborn as sns

df = sns.load_dataset("diamonds")

# Performing First Look Ritual

df.head()
df.info()
df.describe()
```

```python
# Checking how many rows have missing values

count = df.isnull().sum()
print(count)
df
```

```python
# Checking Data types

df.dtypes
```

```python
# Encoding Text Columns

df = pd.get_dummies(df, columns=['cut', 'color', 'clarity'])
df
```

```python
# Train/Test Split

from sklearn.model_selection import train_test_split

x = df.drop(columns=["price"])
y = df["price"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print('Trained Rows: ', len(x_train))
print('Tested Rows: ', len(x_test))
```

```python
# Predicting

from sklearn.linear_model import LinearRegression

lin = LinearRegression()
lin.fit(x_train, y_train)
lin.preds = lin.predict(x_test)
```

```python
# Linear Regression

from sklearn.metrics import mean_squared_error, root_mean_squared_error, r2_score

print('Linear Regression:-')
print(f'MSE: {mean_squared_error(y_test, lin.preds)}')
print(f'RMSE: {root_mean_squared_error(y_test, lin.preds)}')
print(f'R2: {r2_score(y_test, lin.preds)}')
```

```python
# Random Forest

from sklearn.ensemble import RandomForestRegressor

forest = RandomForestRegressor(n_estimators=100, random_state=42)
forest.fit(x_train, y_train)
forest.preds = forest.predict(x_test)

print('Random Forest:-')
print(f'MSE: {mean_squared_error(y_test, forest.preds)}')
print(f'RMSE: {root_mean_squared_error(y_test, forest.preds)}')
print(f'R2: {r2_score(y_test, forest.preds)}')
```

```python
# Important Features

importances = pd.Series(forest.feature_importances_, index=x.columns)
print(importances.sort_values(ascending=False))
```

```python
# Which model wins, and what's the top feature?

print(f'Which model wins: Random Forest')
print(f'Top Feature: carat')
```
