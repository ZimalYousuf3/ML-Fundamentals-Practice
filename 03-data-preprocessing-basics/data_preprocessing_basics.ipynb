```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    'age': [25, 32, np.nan, 45, 51],
    'city': ['Islamabad', 'Karachi', 'Rawalpindi', 'Taxila', 'Murree'],
    'income': [45000, 540000, 610000, np.nan, 890000],
    'bought': [0,1,1,1,0]
})

df
```

```python
# Problem 1: Missing Values ## df.dropna (to delete rows)

print(df.isnull().sum())

df['age'] = df['age'].fillna(df['age'].median())
df['income'] = df['income'].fillna(df['income'].median())

print()

print(df.isnull().sum())
df
```

```python
# Problem 2: Encoding Text Values

df = pd.get_dummies(df, columns = ['city'], drop_first = True)

df
```

```python
# Problem 3: Scaling

from sklearn.preprocessing import StandardScaler

features = df.drop(columns='bought')
target = df['bought']

scaler = StandardScaler()

scaled_features = scaler.fit_transform(features)

print(scaled_features)
```

```python
# Problem 4: Spliting

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)

print('Training Rows:', len(x_train))
print('Testing Rows:', len(x_test))
```
