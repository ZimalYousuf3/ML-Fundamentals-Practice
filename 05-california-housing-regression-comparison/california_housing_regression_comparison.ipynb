```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing

df = fetch_california_housing(as_frame=True).frame

# Performing first look ritual

df.head()
df.info()
df.describe()
```

```python
# Train/Test Split

from sklearn.model_selection import train_test_split

x = df.drop(columns=['MedHouseVal'])
y = df["MedHouseVal"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Predicting

from sklearn.linear_model import LinearRegression

lin = LinearRegression()
lin.fit(x_train, y_train)
lin.preds = lin.predict(x_test)
```

```python
# Linear Regression

from sklearn.metrics import mean_squared_error, root_mean_squared_error, r2_score

MSE = mean_squared_error(y_test, lin.preds)
RMSE = root_mean_squared_error(y_test, lin.preds)
R2 = r2_score(y_test, lin.preds)

print(f'Linear Regression:-')
print(f'MSE: {MSE}')
print(f'RMSE: {RMSE}')
print(f'R2: {R2}')
```

```python
# Random Forest

from sklearn.ensemble import RandomForestRegressor

forest = RandomForestRegressor(n_estimators=100, random_state=42)
forest.fit(x_train, y_train)
forest.preds = forest.predict(x_test)

mse = mean_squared_error(y_test, forest.preds)
rmse = root_mean_squared_error(y_test, forest.preds)
r2 = r2_score(y_test, forest.preds)

print(f'Random Forest:-')
print(f'MSE: {mse}')
print(f'RMSE: {rmse}')
print(f'R2: {r2}')
```

```python
# Important Features

importances = pd.Series(forest.feature_importances_, index=x.columns)
print(importances.sort_values(ascending=False))
```
