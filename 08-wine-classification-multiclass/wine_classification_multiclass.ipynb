```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

data = load_wine(as_frame=True)
x = data.data
y = data.target

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

logreg = LogisticRegression(max_iter=5000)
logreg.fit(x_train, y_train)

forest = RandomForestClassifier(n_estimators=100)
forest.fit(x_train, y_train)

print(f'Logistic Regression Accuracy: {logreg.score(x_test, y_test)}')
print(f'Random Forest Accuracy: {forest.score(x_test, y_test)}')
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

preds = forest.predict(x_test)
cm = confusion_matrix(y_test, preds)
print(cm)
ConfusionMatrixDisplay.from_estimator(forest, x_test, y_test)
plt.show()
```

```python
from sklearn.metrics import classification_report

report = classification_report(y_test, preds, target_names=data.target_names)
print('================ Classification Report ================\n')
print(report)
print('=======================================================')
```
